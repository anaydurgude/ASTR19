```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,2*np.pi,100)
def f(x,a,b):
    return a*np.exp(-1*b*x)
```

```python
a = input("Enter a")
b = input("Enter b")
a=float(a)
b=float(b)
```

```python
y=f(x,a,b)
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel(f"{a}*e^(-{b}x)")
plt.savefig("aexpbx.pdf")
```
