```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0.2,2*np.pi,100)
r = np.linspace(2,0.2,100)
y = np.random.normal(0,r,100)
def f1(x,a,b):
    return a*np.exp(-1*b*x)

def f2(x,a,b):
    return a/(x**b)
```

```python
a = input("Enter a")
b = input("Enter b")
a=float(a)
b=float(b)
```

```python
params1, params_cov1 = curve_fit(f1, x, y)
params2, params_cov2 = curve_fit(f2, x, y)
print(params1, params_cov1, params2, params_cov2)
```

```python
print(f"First function's fit: a = {params1[0]}, b = {params1[1]}")
print(f"Second function's fit: a = {params2[0]}, b = {params2[1]}")
```

```python
plt.scatter(x, y, label='Data')
plt.plot(x, f1(x, params1[0], params1[1]), label='First fit')
plt.plot(x, f2(x, params2[0], params2[1]), label='Second fit')
```
